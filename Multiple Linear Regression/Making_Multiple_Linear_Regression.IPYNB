```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
```

```python
x, y = load_diabetes(return_X_y=True, as_frame=True)
```

```python
x.shape
```

```python
y.shape
```

```python
class mk_mulit_linear_regression:
    def __init__(self):
        self.intsec=None
        self.coff=None

    def fit(self, x_train, y_train):
        x_train=np.insert(x_train,0,1,axis=1)
        beta=np.linalg.inv(np.dot(x_train.T,x_train)).dot(x_train.T).dot(y_train)
        self.intsec=beta[0]
        self.coff=beta[1:]

    def predict(self, X_test):
         y_pric=np.dot(X_test,self.coff)+self.intsec
         return y_pric
    
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
print(x_test.shape)
print(y_test.shape)
```

```python
ml=mk_mulit_linear_regression()
```

```python
ml.fit(x_train,y_train)
```

```python
ml.predict(x_test)
```
